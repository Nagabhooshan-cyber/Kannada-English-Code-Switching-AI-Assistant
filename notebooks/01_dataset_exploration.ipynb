# CoLI-Kanglish data exploration

Run `python scripts/prepare_data.py` from the repository root first. This notebook reads the generated summary and processed train/validation/test JSONL files. The raw dataset is never edited. If no source data is installed, the report records that state instead of fabricating examples.

```python
import json
from pathlib import Path
import pandas as pd

report_path = Path('../reports/data_exploration/coli_kanglish_summary.json')
if not report_path.exists():
    raise FileNotFoundError('Run python scripts/prepare_data.py from the project root first.')
summary = json.loads(report_path.read_text(encoding='utf-8'))
summary
```

```python
if summary.get('status') == 'prepared':
    pd.Series(summary['class_distribution'], name='token_count').sort_values(ascending=False)
else:
    print(summary['message'])
```
